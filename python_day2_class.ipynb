### **Type casting**

```python
# Typecasting helps to change the data type into another data type
# int datatype => change to string,bool,float,complex
# string => can change to int, float ONLY if it looks like a number
# string => "Hi" cannot change to int, float
```

```python
a=100
a1=float(a)              # int => can change to string,bool,float,complex 
print(a1)
a2=bool(a)
print(a2)
a3=str(a)
print(a3)
```

```python
a="hi"
a1=float(a)              # int => can change to string,bool,float,complex 
print(a1)                # Error
a2=bool(a)
print(a2)                # True
a3=int(a)
print(a3)                # Error
```

```python
e=0
e1=bool(e)
print(e1)               # 0 means false in boolean
e2=1 
e3=bool(e2) 
print(e3)               # 1 means true in boolean
e4=" "
e5=bool(e4)
print(e5)               #" " space is the special  character so the output eill be "True"
e6=""
e7=bool(e6) 
print(e7)               #"" it is an empty set  so the  output will be "False"
```

### **Input() Function**

```python
num1=input("Enter the number1:")
num2=input("Enter the number2:")
print(num1+num2)                        # data type was not  given so the given data's concated
```

```python
num1=int(input("Enter the number1:"))
num2=int(input("Enter the number2:"))    # data type was given 
sum=num1+num2
print("sum =",sum)
```

### **Operators**

```python
# Arithmetic operators [+, -, **, /,*, //, %]
print(20 + 4)   # -> 24
print(20 - 4)   # -> 16
print(20 * 4)   # -> 80
print(20 / 4)   # -> 5.0
print(20 ** 4)  # -> 160000
print(20 // 4)  # -> 5  Quotient
print(20 % 4)   # -> 0  Remainder
```

```python
# Assignment operators [=, +=, -=, _=, /=, _*=, //=, %=]
a = 20
print(a)  # 20

a += 5    # a = a + 5 = 25
print(a)  # 25

a *= 2    # a = a * 2 = 50
print(a)  # 50 o/p 50
```

```python
# Comparison operators [==, !=, >, <, >=, <=]
print(20 == 20) # True
print(20 != 20) # False
print(20 > 20)  # False
print(20 < 20)  # False
print(20 >= 20) # True
print(20 <= 20) # True
```
