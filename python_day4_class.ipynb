### **IF ELSE**
* Python checks the condition first, then it checks whether the condition is true or false

```python
if condition:
    # Runs ONLY if condition is True
else:
    # Runs ONLY if condition is False
```

```python
user_message = "hi"

if user_message == "hi":
    print("Hello! How can I help you?")
else:
    print("Sorry, I didn't understand")
```

```python
# Python checks from top to bottom. The moment one condition becomes True, it stops.
#  It will not check the rest.

marks = 85

if marks >= 90:
    print("Excellent, you are topper!")        
elif marks >= 70:
    print("Very good, keep going!")
elif marks >= 40:
    print("Pass, but you can do better")
else:
    print("Fail, don't give up, try again")
```

### **Nested if**

```python
# Nested if => if contains another if

mark=40
if mark >= 35:
    print("Pass")
    if mark >= 90:
        print("First Class")
    elif mark >= 60:
        print("Second Class")
    else:
        print("Third Class")
else:
    print("Fail")
```

### **Short Hand method**

```python
# If the condition is True, the first print will execute. 
# If the condition is False, the second print will execute.

num = 7
print("Even") if num % 2 == 0 else print("Odd")
```

### **While loop**

```python
while <condition>:
    # statements
```

```python
a = 1
while a <= 10:
    print(a)
    a += 2                  #increment by 2
print("Loop over")
```

```python
#Reverse
a = 10
while a >= 1:
    print(a, end=" ")          
    a -= 1
print("\nLoop over")
```
