### **IF ELSE**
* Python checks the condition first, then it checks whether the condition is true or false

```python
if condition:
    # Runs ONLY if condition is True
else:
    # Runs ONLY if condition is False
```

```python
user_message = "hi"

if user_message == "hi":
    print("Hello! How can I help you?")
else:
    print("Sorry, I didn't understand")
```
