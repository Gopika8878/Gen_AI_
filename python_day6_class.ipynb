### **Pattern**

```python
# Pattern Printing using While Loop and For Loop

# while loop
row = int(input("Enter row: "))
col = int(input("Enter col: "))
symbol = input("Enter a symbol: ")
a = 1
while a <= row:
    b = 1
    while b <= col:
        print(symbol, end=" ")
        b += 1
    a += 1
    print()
```

```python
# For Loop 
num=int(input("enter the number"))
symbol=input()
for i in range(num):
    for j in range(0,num+1):
        print(symbol,end=" ")
    print()
```

### **User Defined Function**

```python
def hello():
    print("hi")

hello()

def hello2():                     # You can call a function as many times as you want.
    a = 100
    b = 200
    c = a + b
    print(c)

hello2()
```

### **Function with Parameters**

```python
# Parameter is used to pass a value from outside to inside the function.
def add(x, y):
    print("x : ", x)
    print("y : ", y)
    print("The sum of : ", x + y)        # If you call add(40) only, it will show an error.
                                         # Because both x and y need values.
add(40, 20)                                     
print()


```

### **Function with Default Parameters**

```python
def add_default(x=0, y=0):
    print("x : ", x)
    print("y : ", y)
    print("The sum of : ", x + y)

add_default(40)
print()
```

### **Return keyword**

```python
# return is used to send a value back from a function.
# After return, we can store that value and use it again.

def myfunc(x=0, y=0):
    print("x : ", x)
    print("y : ", y)
    return x + y

ans = myfunc(50, 70)
print(ans)

ans2 = myfunc(50, 50)
print(ans2)

print(myfunc(10, 20))
print()
# Function inside Function
ans = mysub(myadd(10, 10), mysub(20, 10))
print(ans)  # Output: 10
```

```python
 # Using return for calculation
def myadd(x=0, y=0):
    return x + y

def mysub(x=0, y=0):
    return x - y

ans = myadd(10, 50) - mysub(20, 10)
print(ans)

```

```python
# Function inside Function
ans = mysub(myadd(10, 10), mysub(20, 10))
print(ans)  
```

```python
# Factorial - Without Function
mylist = [1, 2, 3, 4, 5]
x = 1
for i in mylist:
    x = x * i
print(x)
print()
```

```python
# Factorial - With Function
mylist=[1, 2, 3, 4, 5]
def my_fact(mylist):
    x = 1
    for i in mylist:
        x = x * i
    return x
ans = my_fact(mylist)
print(ans)  
```

### **Arbitrary Arguments**

```python
# Normal function - takes only 1 value
def myfunc_single(x):
    print(x, type(x))

myfunc_single(10)                             # int
myfunc_single("hai")                          # str
                                              # myfunc_single(10, 30, 55, "hai") -> This will give an error

# *args - stores arguments in a tuple
# You can pass any number of values
def myfunc_args(*x):
    print(x, type(x))

myfunc_args()  # Empty tuple ()
myfunc_args(10, 30, 55, "hai")

# **kwargs - stores keyword arguments in a dict
# You must pass like key = value
def myfunc_kwargs(**x):
    print(x, type(x))

myfunc_kwargs(a=10)
myfunc_kwargs(a=10, b=11, c=5)
myfunc_kwargs(apple=10)
```

### **Function Overloading**

```python
# In Python, if you write same function name many times,
# only the last one will work. The last one will override the others.

def myfunc(x):
    return x

def myfunc(x, y):
    return x + y

def myfunc(x, y, z):
    return x + y + z

print(myfunc(100, 200, 100))  
```
