```python
num=int(input("enter the number"))
symbol=input("enter the symbol")
for i in range(num):
    for j in range(num-i):
        print(symbol,end=" ")
    print()
```

```python
row=int(input("enter the number"))
symbol=input("enter the symbol")
a=1
while a<=row:
    b=1
    while b <= row-a+1:
        print(symbol,end=" ")
        b+=1
    a+=1
    print()
```
