### **Nested Dictionary**

```python
x = {
    "menu": {
        "id": "file",
        "value": "File",
        "popup": {
            "menuitem": [
                {"value": "New", "onclick": "CreateNewDoc()"},
                {"value": "Open", "onclick": "OpenDoc()"},
                {"value": "Close", "onclick": "CloseDoc()"}
            ]
        }
    }
}
print(x.get("menu"))
print(x.get("menu")["popup"])
print(x.get("menu")["popup"]["menuitem"][0])
print(x.get("menu")["popup"]["menuitem"][1]["value"])
```

### **Palindrome**

```python
# Return Palindrome Program

def check_palindrome(mystr):
    newstr = ''
    for each in mystr:
        if each.isalpha():
            newstr = newstr + each.lower()

    if newstr == newstr[::-1]:
        return True
    else:
        return False

print(check_palindrome("Racecar"))
print(check_palindrome("Red Rose"))
print(check_palindrome("MalaYa lAm"))
print(check_palindrome("Never a foot"))
```

### **Character Frequency**

```python
def character_freq(mystr):
    newstr = ''
    for each in mystr:
        if each.isalpha():
            newstr = newstr + each.lower()

    mydict = {}
    for each_1 in newstr:
        if each_1 not in mydict.keys():
            mydict[each_1] = 1
        else:
            mydict[each_1] += 1

    return mydict

print(character_freq("Hello World"))
print(character_freq("Banana"))
```
