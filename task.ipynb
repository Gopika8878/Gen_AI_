```python
for i in range(1,5):
    for j in range(1,5):
        print("*",end=" ")
    print()
```

```python
row = int(input("enter the row"))
col = int(input("enter the column"))
symbol=input("enter the symbol")
a=1
while a<=row:
    b=1
    while b<=col:
        print(symbol,end=" ")
        b+=1
    a+=1
    print()
```

```python
for i in range(5):
    for j in range(0,i+1):
        print("*",end=" ")
    print()
```

```python
row = int(input("enter the row"))
symbol=input("enter the symbol")
i=1
while i<row:
    j=1
    while j<i:
        print(symbol,end=" ")
        j+=1
    i+=1
    print()
```
