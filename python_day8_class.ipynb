### **Replace()**

```python
text = "I hate rain"
result = text.replace("hate", "love")          # replace() changed the word 'hate' to 'love'
print(result)
```

### **count()**

```python
text = "mississippi"
result = text.count("s")                 # letter 's' appears 4 times in 'mississippi'
print(result)
```

### **Find()**

```python
text = "Welcome to Chennai"
result = text.find("to")
print(result)                          # word 'to' starts at index position 8

text = "Welcome to Chennai"
result = text.find("Delhi")
print(result)                         # 'Delhi' is not found, so find() returns -1, no error
```

### **Index()**

```python
text = "Welcome to Chennai"
result = text.index("Chennai")
print(result)                         # word 'Chennai' starts at index position 11
```

### **Format()**

```python
name = "Kavin"
city = "Salem"
print("Hi, I am {} from {}".format(name, city))   # first {} took 'Kavin', second {} took 'Salem'

print("Hi, I am {1} from {0}".format(city, name)) #{0} means first value 'Salem', {1} means second value 'Kavin'

order = "I ordered {item} in {color} colour"
print(order.format(item="phone", color="blue"))   # named placeholders filled with given values
```

### **String Formatting**
* **%d - Integer**
* **%s - String**
* **%f - Float with default 6 decimal places**
* **%.2f - Float with two decimal places**
* **%.10f - Float with ten decimal places**

```python
name = "Sara"
age = 19
score = 88.765
print("%s is %d years old, score is %.2f" % (name, age, score))  # %s = string, %d = integer, 
                                                                 # %.2f = float with 2 decimal places, 
                                                                 # 88.765 rounded to 88.77
```

```python
fruit = "mango"
price = 60
quantity = 2
print(f"I bought {quantity} {fruit} for {price * quantity} rupees")# {price * quantity} calculated 60 * 2 = 120 
```

### **Dictionary Function**

```python
mydict = {"Name": "Sara", "City": "Madurai", "Course": "Data Science"}
print(mydict["Course"])                           # It prints value of key 'Course'

mydict = {"Name": "Sara", "Course": "Data Science", "Course": "AI"}
print(mydict["Course"])                           # Same key repeated, so last value 'AI' overwrites first value

# Add new element
mydict["Age"] = 25
print(mydict)                                     # New key 'Age' added to dictionary
```

```python
# keys() - Returns only keys
mydict = {"Name": "Sara", "Course": "AI"}
print(mydict.keys())                               # keys() returns only keys

print(list(mydict.keys()))                         # list() converted keys to list

for each in mydict.keys():
    print(each)                                    # for loop prints one by one
```

```python
# values() - Returns only values
mydict = {"Name": "Sara", "Course": "AI"}
print(mydict.values())                              # values() returns only values

for each in mydict.values():
    print(each)
```

```python
# items() - Key and Value together as Tuple
mydict = {"Name": "Sara", "Course": "AI"}
print(mydict.items())                             # items() returns key + value as tuple pair

for each in mydict.items():
    print(each)                                   # each pair printed as tuple

for each in mydict.items():
    print(f"{each[0]} : {each[1]}")               # each[0] is key, each[1] is value
```

```python
# pop() - List, Set, Dict

# pop() in List
# Two types -> pop(), pop(index)
x = [10, 20.5, "hi", 10, "hello"]
print(x)
x.pop()
print(x)                                          # pop() removes last element 'hello'
x.pop(2)
print(x)                                          # pop(2) removes item at index 2, which is 'hi'

# pop() in Set - Random delete
x = {10, 20.5, 10, 21}
x.pop()
print(x)                                          # Set has no order, so pop() removes random item

# pop(key) in Dict
mydict = {"Name": "Sara", "Course": "Python"}
mydict.pop("Course")
print(mydict)                                     # pop("Course") deleted the key 'Course'
```

```python
# popitem() - Deletes last item
mydict = {"Name": "Sara", "Course": "Python", "Fee": 500}
mydict.popitem()
print(mydict)                                       # popitem() removes last inserted item 'Fee'
```

```python
# clear() - Deletes everything
mydict = {"Name": "Sara", "Course": "Python", "Fee": 500}
print(mydict)
mydict.clear()
print(mydict)
```

```python
# update() - Update already existing dict
mydict = {"Name": "Sara", "Course": "Python"}
newdict = {"Name": "Meena", "Course": "AI", "Fee": 401, "Age": 20}
mydict.update(newdict)
print(mydict)                                          # update() changed old values and added new keys
```

```python
# get() - To get value
mydict = {"Name": "Sara", "Course": "Python"}
print(mydict.get(600))                                # Key 600 is not there, so get() returns None, no error

print("Hello Python")                                 # Program continues, it did not stop for error
```

```python
# copy() - Copy to different location
x = {"A": 100, "B": 200}
y = x.copy()
x["A"] = 400
y["E"] = 500
print(x)                                               # x changed only, y not affected
print(y)                                               # y is separate copy, so new key 'E' added only in y
```

```python
# Nested Dictionary
mydict = {
    "Name": "Sara",
    "City": ["Madurai", "Trichy", "Mumbai"],
    "Year": [2000, 2010, 2020]
}

print(mydict.get("City"))                                # Value of 'City' is a full list

print(mydict.get("City")[0])                             # [0] took first city from list

print(mydict.get("City")[2][3:])                         # City[2] is 'Mumbai', [3:] took from index 3, so 'bai
```
