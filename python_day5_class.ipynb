### **BREAK - Stop the loop**

```python
a = 1
while a <= 10:
    print(a, end=" ")
    a += 1
print("Loop over")
```

```python
# Stop when a reaches 20
a = 1
while a <= 10:
    if a >= 20:
        break
    print(a, end=" ")
    a += 1
```

### **CONTINUE - Go back to the top of the loop**

```python
a = 1
while a <= 10:
    if a % 3 == 0:
        a += 1
        continue                      # Skip printing numbers divisible by 3
    print(a, end=" ")
    a += 1
print("\nLoop over")                  # Output: 1 2 4 5 7 8 10
                                      # 3, 6, 9 are skipped
```

### **NESTED WHILE LOOP**

```python
a = 1                                  # row
while a <= 3:
    b = 1                              #col
    while b <= 5:
        print(b, end=" ")
        b += 1
    a += 1
    print()                            # Next line
```

```python
row = int(input("Enter row: "))
col = int(input("Enter col: "))

a = 1
while a <= row:
    b = 1
    while b <= col:
        print(b, end=" ")
        b += 1
    a += 1
    print()
```

### **FOR LOOP**
* **Syntax: for <variable_name> in <Str, List, Tuple, dict, Set>**

```python
for x in "Hello world":
    print(x,end="")
```

```python
for x in range(91, 100):
    print(x, end=" ")
print()
```

```python
# range(start, stop, jump)
for x in range(0, 11, 2):
    print(x, end=" ")
print()
```

```python
# reverse
for x in range(10, 0, -2):
    print(x, end=" ")
print()
```

### **FOR WITH ELSE**
* **If for loop completes successfully, else will run**

```python
for x in "Hello world":
    print(x, end=" ")
else:
    print("\nCompleted")
```

```python
# For with Else + Break
# If break happens, else will NOT run

for x in "Hello World":
    if x == " ":
        break                                          # Stop when space comes
    print(x, end="")
else:
    print("Completed")
```

```python
mylist = [2, 6, 9, 11, 16, 97]
for each in mylist:
    if each % 2 == 0:
        print(each, "Divisible by 2")
        break
else:
    print("Number divisible by 2 not found")
```

### **NESTED FOR LOOP**

```python
for x in range(5):
    for y in range(1, 6):
        print(y, end=" ")
    print()
```
