### **LIST - [ ]**

```python
# Ordered 
# Mutable 
# Allows duplicate values

mylist = [10, 30, 55, "Hello", "hi"]     # Print the value at index 2
print(mylist[2])                         # Output: 55

mylist[2] = "hi"                         # List is changeable, so we can change index 2 value
print(mylist)                            # Output: [10, 30, 'hi', 'Hello', 'hi']
```

### **TUPLE-()**

```python
# Ordered
# Immutable (we CANNOT change the values)
# Allows duplicate values

mytuple = (10, 30, 50, "hi", "hello")

print(mytuple[1])                       # For index we must use [ ], not ( )        
                                        # Output: 30

# Trying to change a Tuple will give an Error
# mytuple[2] = "hi"                     # Error: 'tuple' object does not support item assignment

```

### **SET-{}**

```python
# Unordered 
# Unindexed
# Mutable (we can add / remove values)
# Does NOT allow duplicate values

# Do not confuse Set and Dict
# {"hi": "hello"} -> This is a Dict (Key : Value)
# {"hi", "hello", 40} -> This is a Set (only values)

myset = {"hi", "hello", 40, 50, 40}
print(myset)                                     # Output: {'hi', 'hello', 40, 50}
                                                 # Duplicate is not allowed, so 40 will appear only one time
```

### **DICT-{Key:value}**

```python
# Unordered, Unindexed
# Mutable (we can change the values)
# Does NOT allow duplicate KEYS but allows duplicate VALUES

mydict = {"Name": "Gopi", 500: 200, "Age": 22}
print(mydict)                                     # Output: {'Name': 'Gopi', 500: 200, 'Age': 22}

# We can change the value using the Key
mydict[500] = "welcome"
print(mydict)                                     # Output: {'Name': 'Gopi', 500: 'welcome', 'Age': 22}
```

### **MUTABLE vs IMMUTABLE OBJECTS**

```python
# Immutable -> int, float, bool, str, tuple
# Mutable -> list, set, dict

# Immutable:
# Same data stores in SAME location
# Different data stores in DIFFERENT location

# Mutable:
# Same data stores in DIFFERENT location
# Different data stores in DIFFERENT location
```

### **Immutable**

```python
# Create a tuple
fruits = ("apple", "banana", "mango")
print(fruits)  # Output: ('apple', 'banana', 'mango')

# CHANGE a value => This will give ERROR
fruits[1] = "orange"                           # Error: 'tuple' object does not support item assignment

# We cannot change it. We have to create a completely new tuple.
```

### **Mutable**

```python
# Create a list
fruits = ["apple", "banana", "mango"]
print(fruits)                                    # Output: ['apple', 'banana', 'mango']

# CHANGE a value
fruits[1] = "orange"
print(fruits)                                    # Output: ['apple', 'orange', 'mango']
```

### **Identity Operators => is ,is not**

```python
# It returns True or False only
# It checks not only the value, but also the address
# "is" means  check the address 

a = 100
b = 100
print(a == b)                               # True - value is same
print(a is b)                               # True - same location

a = [100, 200]
b = [100, 200]
print(a == b)                               # True - value is same
print(a is b)                               # False - different location

a = 100
b = 100
print(a == b)                               # True
print(a is not b)                           # False - because a and b are same

a = [100, 200]
b = [100, 200]
print(a == b)                               # True
print(a is not b)                           # True - because location is different
```

### **Membership operators => in, not in** 

```python
a = [100, 200]
b = [400, 300]

print(100 in a)  # True - 100 is inside a
print(300 in b)  # True - 300 is inside b

a = [100, 200]
b = [400, 300]

print(100 not in b)  # True - 100 is not inside b
print(100 not in a)  # False - 100 is inside a
```
